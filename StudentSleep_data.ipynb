```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
# Data of Student Sleep, Screen Time & Mental Health 2026
Student_data = pd.read_csv("student_sleep_mental_health_2026.csv")
Student_data = Student_data.drop_duplicates()
print(Student_data.head())
```

```python
# This charts shows how sleep and stress_level are connected

sns.lineplot( 
    data = Student_data,
    x= "avg_sleep_hours", y = "stress_level", hue="feels_burned_out"
)
```

```python
# This chart shows how Sleep can effect Anxity

sns.lineplot(data=Student_data, x='avg_sleep_hours', y='anxiety_score')
plt.title("Anxiety Score by Sleep Quality")
plt.show()
```

```python
# Avrage Sleep time

sns.histplot(Student_data['avg_sleep_hours'], kde=True, bins=20)
plt.title("Distribution of Sleep Duration")
plt.show()
```

```python
# Age and stress_level

sns.lineplot(
    data = Student_data,
    x = "age", y = "stress_level"
)
```

```python
# These charts shows how screen_time and stress_level are connected
sns.boxenplot(
    data = Student_data,
    x= "stress_level", y= "screen_time_hours"
)
plt.show()
sns.barplot(
     data = Student_data,
    x= "stress_level", y= "screen_time_hours", hue="gender"
)
```

```python
# This chart shows that students use phones mainly for social_media

sns.lineplot(
    data= Student_data,
    x = "screen_time_hours", y = "social_media_hours", color = "green"
)
```

```python
sns.lineplot(
    data = Student_data,
    x= "screen_time_hours", y= "study_hours_per_day", color = "black"
)
```

```python
sns.scatterplot(data=Student_data, x='avg_sleep_hours', y='stress_level', hue='gender')
plt.title("Sleep Duration vs Stress Level")
plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(Student_data.select_dtypes(include='number').corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.pairplot(
    Student_data[
        [
            'avg_sleep_hours',
            'stress_level',
            'anxiety_score',
            'study_hours_per_day',
            'screen_time_hours',
            'feels_burned_out'
        ]
    ],
    hue='feels_burned_out',
    diag_kind='hist'
)

plt.show()
```

```python
# Select only numerical columns
numeric_data = Student_data.select_dtypes(include=['number'])

# Correlation matrix
corr = numeric_data.corr()

# Create the heatmap
plt.figure(figsize=(12, 8))

sns.heatmap(
    corr,
    annot=True,          # Show correlation values
    cmap='coolwarm',     # Blue = negative, Red = positive
    fmt='.2f',           # Show 2 decimal places
    linewidths=0.5,
    square=True
)

plt.title("Correlation Heatmap of Student Mental Health Dataset", fontsize=16)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)

plt.tight_layout()
plt.show()
```

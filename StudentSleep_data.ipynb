```python
"""
Radar / Polygon Chart — Student Lifestyle Profile by Sleep Category
---------------------------------------------------------------------
Compares Low / Medium / High sleep groups across 7 lifestyle &
mental-health metrics on one polygon chart.
"""

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
CALM = ["#264653", "#2A9D8F", "#8AB17D", "#E9C46A", "#F4A261", "#E76F51"]

# ---------------------------------------------------------------
# 1. LOAD + GROUP
# ---------------------------------------------------------------
students_data = pd.read_csv("student_sleep_mental_health_2026.csv")  # <- update path if needed

students_data['sleep_category'] = pd.cut(
    students_data['avg_sleep_hours'], bins=[0, 6, 7.5, 24],
    labels=['Low (<6h)', 'Medium (6-7.5h)', 'High (>7.5h)']
)

metrics = ['avg_sleep_hours', 'screen_time_hours', 'social_media_hours',
           'exercise_hours_per_week', 'stress_level', 'anxiety_score', 'gpa']

labels = ['Sleep\nHours', 'Screen\nTime', 'Social\nMedia',
          'Exercise', 'Stress', 'Anxiety', 'GPA']

group_order = ['Low (<6h)', 'Medium (6-7.5h)', 'High (>7.5h)']
group_means = students_data.groupby('sleep_category', observed=True)[metrics].mean().reindex(group_order)

# ---------------------------------------------------------------
# 2. NORMALIZE (0-1) using the full dataset's range for each metric
#    so the polygons are fairly comparable across axes
# ---------------------------------------------------------------
norm = (group_means - students_data[metrics].min()) / (students_data[metrics].max() - students_data[metrics].min())

# ---------------------------------------------------------------
# 3. BUILD THE RADAR CHART
# ---------------------------------------------------------------
n = len(metrics)
angles = np.linspace(0, 2 * np.pi, n, endpoint=False).tolist()
angles += angles[:1]  # close the polygon

fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))
ax.set_theta_offset(np.pi / 2)
ax.set_theta_direction(-1)

colors = [CALM[5], CALM[4], CALM[1]]  # red -> orange -> teal (low -> high sleep)

for i, group in enumerate(group_order):
    values = norm.loc[group].tolist()
    values += values[:1]
    ax.plot(angles, values, linewidth=2.2, label=group, color=colors[i])
    ax.fill(angles, values, alpha=0.15, color=colors[i])

ax.set_xticks(angles[:-1])
ax.set_xticklabels(labels, fontsize=11, fontweight='bold')
ax.set_yticklabels([])
ax.set_ylim(0, 1)
ax.spines['polar'].set_visible(False)
ax.grid(color='grey', alpha=0.3)

plt.title("Student Lifestyle Profile by Sleep Category", fontsize=15,
          fontweight='bold', pad=30)
plt.legend(loc='upper right', bbox_to_anchor=(1.35, 1.1), title="Sleep Group",
           frameon=False)

plt.figtext(0.5, -0.02,
            "Values scaled 0-1 across the full dataset range per metric.\n"
            "Note: Stress/Anxiety are 'bad' when high; GPA/Exercise/Sleep are 'good' when high.",
            ha='center', fontsize=9, color='gray')

plt.tight_layout()
plt.savefig("radar_sleep_category.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
# Data of Student Sleep, Screen Time & Mental Health 2026
Student_data = pd.read_csv("student_sleep_mental_health_2026.csv")
print(Student_data.head())
```

```python
sns.lineplot( 
    data = Student_data,
    x= "avg_sleep_hours", y = "stress_level", hue="feels_burned_out"
)
```

```python
sns.histplot(Student_data['avg_sleep_hours'], kde=True, bins=20)
plt.title("Distribution of Sleep Duration")
plt.show()
```

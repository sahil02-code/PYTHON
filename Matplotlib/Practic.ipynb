```python
# Movies Data

# Oscar Winner Movies
oscar_movies = [
    "The Dark Knight", 
    "The Hurt Locker",  
    "The King's Speech", 
    "The Artist",     
    "Argo"
]
oscar_years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M
plt.plot(oscar_years, oscar_revenue)
plt.title("Oscar Winner Movies Revenue")
plt.xlabel("years")
plt.ylabel("revenue in $M")
```

```python
# Plot multiple
# Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_years = [2008, 2009, 2010, 2011, 2012]
non_oscar_revenue = [378, 2788, 829, 185, 275]

plt.plot(oscar_years, oscar_revenue,label = "oscar winners")
plt.plot(non_oscar_years,non_oscar_revenue,label = "non-oscar winners")
plt.title("Oscar Winner Movies Revenue")
plt.xlabel("years")
plt.ylabel("revenue in $M")
plt.legend()
```

```python
# Format stings
plt.plot(oscar_years, oscar_revenue,"p:b",label = "oscar winners")
plt.plot(non_oscar_years,non_oscar_revenue,"o--r",label = "non-oscar winners")

plt.plot(oscar_years, oscar_revenue, color="black", linestyle="-", marker="o",label="Oscar Winners")
plt.plot(non_oscar_years, non_oscar_revenue, color="red", linestyle="--", marker=">", linewidth=0, label="Non-Oscar Winners")
plt.legend()
```

```python
# Other Methods & Attributes
print(plt.style.available)
plt.plot(oscar_years, oscar_revenue, label="Oscar Winners")
plt.style.use('dark_background')
plt.style.use('ggplot')

#To show square grid lines in background
plt.grid(True)

with plt.xkcd():
    plt.plot(oscar_years, oscar_revenue, "o-b", label="Oscar Winners")
    plt.plot(non_oscar_years, non_oscar_revenue, ">--g", label="Non-Oscar Winners")
    plt.xlabel("Years")
    plt.ylabel("Revenue(in $M)")
    plt.legend()
plt.style.available
```

```python
#Showing multiple bars in a single chart

import numpy as np

oscar_movies = [
    "The Dark Knight", 
    "The Hurt Locker",  
    "The King's Speech", 
    "The Artist",     
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M

non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]

x = np.arange(len(years))
width = 0.4

plt.title("oscar winning movies vs non_socar movies revenue")
plt.xlabel("Years")
plt.ylabel("Revenue in (M$)")

plt.bar(x-width/2, oscar_revenue, width, label = "oscar_movies")
plt.bar(x+width/2, non_oscar_revenue , width,  label = "non_oscar_movies")
plt.xticks(x, years) # to show x-axis labels
```

```python
#Scatter plot 

people = ["Person A", "Person B", "Person C", "Person D", "Person E", 
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
blood_pressure = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]
#plt.scatter(age, blood_pressure)

plt.title("age to blood_pressure")
plt.xlabel("age")
plt.ylabel("blood_pressure")
plt.grid()

# Customizing plots
color = ["green" if bp> 135 else "red" for bp in blood_pressure ]
sizes = blood_pressure
plt.scatter(age, blood_pressure, s = sizes ,color = color, alpha= 0.8) #alpha = transpirensi

#cmap = color maps and c = which list is going to get the color
# plt.scatter(age, blood_pressure, c=blood_pressure, cmap="plasma")
plt.scatter(age, blood_pressure, c=blood_pressure, cmap="OrRd")# makes more sense
# plt.scatter(age, blood_pressure, c=[5]*len(age), cmap="plasma")
plt.colorbar(label="BP")

for i in range (len(age)):
    plt.annotate(people[i], xy =(age[i], blood_pressure[i]),xytext =(age[i]+2, blood_pressure[i]+2))
plt.xlim(20, max(age)+10)
```

```python
#Pie chart
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]

plt.pie(amounts, labels = expenses)
plt.style.available
plt.style.use("dark_background")

# add %-age of each slice
plt.pie(amounts, labels= expenses, autopct="%1.2f%%")

#set color
color = ["pink","blue","purple","green", "yellow"]
plt.pie(amounts, labels= expenses, autopct="%1.2f%%", colors= color)

# wedgeprops
plt.pie(amounts, labels= expenses, autopct="%1.2f%%",
        wedgeprops={"edgecolor": "white",
                    "linewidth" : 1,
                    "linestyle": ":",
                    "fill": False,
                    "animated": True}
        )
```

```python
# Explode
explode = [0,0,0,0, 0.2]
plt.pie(amounts, labels= expenses, explode = explode)

# shadow
plt.pie(amounts, labels= expenses, explode = explode, shadow= True)
plt.style.use("classic")

# Angle
# plt.pie(amounts, labels= expenses, explode = explode, shadow= True, startangle= 90)
#top
# plt.pie(amounts, labels= expenses, explode = explode, shadow= True, startangle= 180)
```

```python
#Histograms

#generate random data
import numpy as np

np.random.seed(0)
#generates 100 nums with mean 70 and std dev 10
score = np.random.normal(loc= 70, scale= 10, size= 1000)

plt.hist(score, bins= 20, color="skyblue", edgecolor= "black")
plt.title("Distribution of exam scores")
plt.xlabel("Score")
plt.ylabel("Number of students")
```

```python
import numpy as np

np.random.seed(0)
#generates 100 nums with mean 70 and std dev 10
score = np.random.normal(loc= 70, scale= 10, size= 1000)

plt.hist(score, bins= 20, color="skyblue", edgecolor= "black")
plt.title("Distribution of exam scores")
plt.xlabel("Score")
plt.ylabel("Number of students")

# Mean marks
plt.axvline(
    x= np.mean(score),
    color= "green",
    linewidth = 1,
    linestyle = "-.",
    label = (f"Mean marks {np.mean(score): .2f}")
)
plt.legend()
```

```python
# Multiple datasets
legit_transactions = [
    2.99, 5.49, 8.99, 12.50, 14.99, 19.99, 23.45, 29.99, 34.99, 39.50,
    45.00, 49.99, 55.25, 60.00, 75.99, 89.99, 120.50, 150.00, 199.99,
    249.99, 300.75, 450.00, 600.00, 850.00, 1200.00
]

fraud_transactions = [
    50, 100, 150, 200, 300, 500, 500, 750, 1000, 1000,
    1200, 1500, 1500, 2000, 2500, 3000, 3000
]

plt.hist(fraud_transactions, bins= 10, label="Fraud", color= "red", alpha = 0.5, edgecolor = "black")

plt.hist(legit_transactions, bins= 10, label="legit", color= "blue", alpha = 0.5, edgecolor = "black")

plt.title("Transection Amount Distribution: Legitimate VS Fraudulent")
plt.xlabel("Transection Amounts ($)")
plt.ylabel("Frequency")
plt.legend()
plt.grid()
```

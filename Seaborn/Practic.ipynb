```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Q1
import numpy as np
data = np.random.normal(loc=70, scale=10, size=200)

fig, axes = plt.subplots()
mean_value = np.mean(data)
axes.hist(data, bins = 70)
axes.axvline(x = mean_value, color = "red", linestyle = "-.", linewidth = 2,label = f"mean = {mean_value:2f}")
axes.set_xlabel("Score")
axes.set_ylabel("Frequency")
axes.set_title("Histogram of Scores")
axes.legend()
plt.show()
```

```python
# Q3

tips = sns.load_dataset("tips")
print(tips.head())

sns.boxplot(
    data = tips,
    x = "day", y = "total_bill",
    hue = "sex", showmeans = True
)
```

```python
# Q4

tips = sns.load_dataset("tips")
sns.scatterplot(
    data = tips,
    x= "total_bill", y = "tip",
    hue = "sex", style="smoker"
)
plt.grid()
plt.title("Bill VS Tip")
plt.show()
```

```python
# Q5
tips = sns.load_dataset("tips")
print(tips.head())

# Create pivot table
pivot = tips.pivot_table(
    values="tip",
    index="day",
    columns="time",
    aggfunc="mean"
)
print(pivot)

sns.heatmap(
    data = pivot,
    annot=True
)
```

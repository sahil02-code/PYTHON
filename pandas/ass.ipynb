```python
data = pd.read_csv("IRIS.csv")
result = data[(data["petal_length"] > 4.5) & (data["species"] == "Iris-virginica")]

print(result)
```

```python
data.groupby("species").agg({
    "sepal_length": "mean",
    "petal_width": "max",
    "sepal_width": "std"
})
```

```python
data["petal_ratio"] = data["petal_length"] / data["petal_width"]
data.head()
data.groupby("species")["petal_ratio"].mean()
```
